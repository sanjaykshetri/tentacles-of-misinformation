# Election Misinformation Case Study: Analytical Framework

## Overview

This notebook structures the analysis of AI-generated election misinformation across 6 regions using a consistent 4-dimension framework:

1. **Modality**: What is it? (text, audio, video, or combination?)
2. **Speed of Spread**: How fast did it propagate? Platform dynamics?
3. **Cognitive Triggers**: Which behavioral vulnerabilities (from Chapter 1) did it exploit?
4. **Detectability**: Would existing systems catch it? What new signals help?

---

## Regions Covered

- 🇸🇰 Slovakia 2023 (deepfake audio)
- 🇮🇳 India 2024 (AI-generated speeches)
- 🇺🇸 United States (robocall deepfake)
- 🇳🇵 Nepal 2024 (social media deepfakes)
- 🌏 South Asia broader patterns
- 🌎 Latin America (video deepfakes)

---

## Framework 1: Modality Analysis

What modalities are being used, and how prevalent is each?

```python
# Extract and visualize modality breakdown
modality_data = cases_df['modality'].value_counts()

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Modality frequency
modality_data.plot(kind='bar', ax=axes[0], color=['#e74c3c', '#3498db', '#2ecc71'])
axes[0].set_title('Modality Distribution Across Cases', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Modality')
axes[0].set_ylabel('Count')
axes[0].tick_params(axis='x', rotation=45)

# Platform reach by modality
platform_reach = cases_df.groupby('modality')['estimated_reach'].sum().sort_values(ascending=False)
platform_reach.plot(kind='barh', ax=axes[1], color=['#e74c3c', '#3498db', '#2ecc71'])
axes[1].set_title('Total Estimated Reach by Modality', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Estimated Reach (cumulative)')
plt.tight_layout()
plt.show()

print(f"\nModality Breakdown:\n{modality_data}")
print(f"\nTotal Reach by Modality:\n{platform_reach}")
```

## Framework 2: Speed of Spread Analysis

Cascade dynamics (R₀ estimates, doubling time, platform effects)

```python
# Analyze R0 and cascade dynamics
r0_data = cases_df[['region', 'r0_estimate', 'platform_primary']].copy()
r0_data = r0_data.sort_values('r0_estimate', ascending=False)

print("R₀ Estimates (Basic Reproduction Number) by Region:")
print("----------------------------------------------")
for idx, row in r0_data.iterrows():
    interpretation = (
        'Very low (limited viral potential)' if row['r0_estimate'] < 0.5
        else 'Low (single-digit cascade)' if row['r0_estimate'] < 1.0
        else 'Moderate (self-sustaining growth)' if row['r0_estimate'] < 1.5
        else 'High (exponential growth)'
    )
    print(f"{row['region']:20s} | R₀ = {row['r0_estimate']:.2f} ({interpretation})")

# Visualize
fig, ax = plt.subplots(figsize=(10, 6))
y_pos = np.arange(len(r0_data))
colors = ['#e74c3c' if x >= 1.0 else '#f39c12' if x >= 0.5 else '#3498db' for x in r0_data['r0_estimate']]
ax.barh(y_pos, r0_data['r0_estimate'], color=colors)
ax.set_yticks(y_pos)
ax.set_yticklabels(r0_data['region'])
ax.axvline(x=1.0, color='red', linestyle='--', linewidth=2, label='Epidemic Threshold (R₀=1.0)')
ax.set_xlabel('R₀ Estimate')
ax.set_title('Cascade Dynamics: Estimated R₀ by Region', fontweight='bold')
ax.legend()
plt.tight_layout()
plt.show()

print(f"\nAverage R₀: {cases_df['r0_estimate'].mean():.2f}")
print(f"R₀ Range: {cases_df['r0_estimate'].min():.2f} - {cases_df['r0_estimate'].max():.2f}")
```

## Framework 3: Cognitive Triggers Mapping

Which behavioral vulnerabilities from Chapter 1 do these cases exploit?

```python
# Define cognitive triggers for each case
cognitive_triggers = {
    'Slovakia_2023': {
        'authority_bias': 0.9,  # mimics candidate's voice
        'urgency': 0.95,  # released days before election
        'tribal_identity': 0.85,  # coded as anti-progressive
        'conspiracy_resonance': 0.8,  # election rigging narrative
        'emotional_arousal': 0.75,  # anxiety about election integrity
    },
    'India_2024': {
        'authority_bias': 0.85,  # attributed to political leaders
        'tribal_identity': 0.95,  # regional + caste targeting
        'familiarity': 0.9,  # regional languages, cultural refs
        'outrage': 0.8,  # controversial political positions
        'scale_effect': 0.85,  # normalization through ubiquity
    },
    'United_States_2024': {
        'authority_bias': 0.95,  # perfect Biden voice
        'action_suppression': 0.9,  # voter suppression framing
        'distrust_system': 0.85,  # election compromise subtext
        'recency': 0.95,  # immediate before primary
        'urgency': 0.9,  # time-sensitive action
    },
    'Nepal_2024': {
        'tribal_identity': 0.9,  # ethnic/caste divisions
        'authority_bias': 0.75,  # candidate attacks
        'emotional_arousal': 0.85,  # compromising scenarios
        'conspiracy_resonance': 0.8,  # lack of fact-check infrastructure
        'distrust_system': 0.85,
    },
}

# Create heatmap
trigger_df = pd.DataFrame(cognitive_triggers).T

fig, ax = plt.subplots(figsize=(12, 6))
sns.heatmap(trigger_df, annot=True, fmt='.2f', cmap='RdYlGn', vmin=0, vmax=1, 
            cbar_kws={'label': 'Trigger Strength'}, ax=ax, linewidths=0.5)
ax.set_title('Cognitive Vulnerability Exploitation by Case', fontweight='bold', fontsize=12)
ax.set_xlabel('Cognitive Trigger')
ax.set_ylabel('Case')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

print("Cognitive Trigger Analysis:")
print("============================\n")
for case, triggers in cognitive_triggers.items():
    avg_trigger = np.mean(list(triggers.values()))
    top_trigger = max(triggers, key=triggers.get)
    print(f"{case}:")
    print(f"  Average trigger strength: {avg_trigger:.2f}")
    print(f"  Dominant trigger: {top_trigger} ({triggers[top_trigger]:.2f})")
    print()
```

## Framework 4: Detectability Assessment

Would existing systems catch these? What new signals help?

```python
# Define detectability profiles for each case
detectability_profiles = {
    'Slovakia_2023': {
        'text_nLP': 0.0,  # no text artifact
        'audio_forensics': 0.55,  # baseline audio detection
        'video_deepfake': None,  # N/A
        'metadata': 0.4,  # upload source suspicious
        'infrastructure_monitoring': 0.6,  # platform anomaly detection
        'ensemble': 0.55,  # combined signal
    },
    'India_2024': {
        'text_nLP': 0.5,  # if transcribed
        'audio_forensics': 0.5,  # mixed quality
        'video_deepfake': 0.45,
        'metadata': 0.8,  # coordinated timestamps
        'infrastructure_monitoring': 0.75,  # organizational pattern
        'ensemble': 0.8,  # pattern detection wins
    },
    'United_States_2024': {
        'text_nLP': 0.0,  # no text
        'audio_forensics': 0.5,  # challenging robocall audio
        'video_deepfake': None,  # N/A
        'metadata': 0.85,  # calling pattern anomaly
        'infrastructure_monitoring': 0.9,  # FCC reporting
        'ensemble': 0.75,  # infrastructure detection critical
    },
    'Nepal_2024': {
        'text_nLP': 0.65,  # Nepali language model gap
        'audio_forensics': 0.3,  # limited Nepali training data
        'video_deepfake': 0.25,  # limited South Asian face datasets
        'metadata': 0.5,
        'infrastructure_monitoring': 0.35,  # limited platform resources
        'ensemble': 0.5,  # gap widens
    },
}

# Visualize detection effectiveness
det_df = pd.DataFrame(detectability_profiles).T

fig, ax = plt.subplots(figsize=(12, 6))
det_df.plot(kind='bar', ax=ax, width=0.8)
ax.set_title('Detection Effectiveness by System and Case', fontweight='bold', fontsize=12)
ax.set_xlabel('Case')
ax.set_ylabel('Detection Likelihood (0-1)')
ax.set_ylim(0, 1)
ax.axhline(y=0.5, color='red', linestyle='--', linewidth=1, alpha=0.5, label='50% threshold')
ax.legend(loc='upper left', fontsize=9)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

print("Detectability Summary:")
print("=====================\n")
for case, det_profile in detectability_profiles.items():
    ensemble_score = det_profile.get('ensemble', 0)
    print(f"{case}:")
    print(f"  Ensemble detectability: {ensemble_score:.0%}")
    print(f"  Strongest signal: {max(det_profile, key=lambda k: det_profile[k] if det_profile[k] is not None else 0)}")
    print()
```
